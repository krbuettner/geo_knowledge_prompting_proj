### Helper Code to Preprocess DollarStreet 

```python
import os.path
import pickle 
import PIL 
from PIL import Image 
```

```python
DS_IMG_OF_INTEREST_PATH = '/ihome/akovashka/krb115/projects/geo_knowledge_prompting_proj/GeoKnowledgePrompting/data_loc/dollarstreet/dollarstreet_img_set.pkl'
PATH_TO_ORIG_KAGGLE_DATA = '/ix/akovashka/krb115/datasets/DollarStreet/dataset_dollarstreet/'
PATH_TO_PROC_DATA_PATH = '/ix/akovashka/krb115/datasets/DollarStreet/dataset_dollarstreet/images_of_interest_preproc/'
```

```python
with open(DS_IMG_OF_INTEREST_PATH, 'rb') as f:
    img_data = pickle.load(f)
```

```python
for i, entry in enumerate(img_data):
    if i % 200 == 0:
        print(i)

    img_id = entry[0]
    prev_img_path = entry[5]

    if not os.path.isfile(PATH_TO_PROC_DATA_PATH + img_id + ".jpg"):
        image = Image.open(os.path.join(PATH_TO_ORIG_KAGGLE_DATA, prev_img_path))
        image = image.resize((256,256))
        image.save(PATH_TO_PROC_DATA_PATH + img_id + ".jpg")
print(img_id)
print(prev_img_path)
```
